```python
import os
import torch
from TTS.api import TTS


device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")


os.environ["COQUI_TOS_AGREED"] = "1"


print("Downloading and loading XTTS-v2 (this takes 1-2 minutes the first time)...")
tts = TTS("tts_models/multilingual/multi-dataset/xtts_v2").to(device)


text_to_speak = "This is a live test of a custom deep learning voice cloning pipeline running on a remote server. The architecture has successfully matched my vocal signature."
reference_audio = "memphis-acapella-phonk-who-the-f-k_158bpm_A#_minor.wav"
output_path = "cloned_output.wav"


print("Synthesizing speech... please wait...")
tts.tts_to_file(
    text=text_to_speak,
    speaker_wav=reference_audio,
    language="en",
    file_path=output_path
)

print(f"Done! Audio file generated successfully at: {output_path}")
```
